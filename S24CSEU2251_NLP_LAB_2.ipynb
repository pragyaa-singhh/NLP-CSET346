```python
import nltk
from nltk.corpus import wordnet as wn

# Download WordNet resources
nltk.download('wordnet')
nltk.download('omw-1.4')

print("WordNet loaded successfully!")
```

```python
# Select 10 English words
words = [
    'car',
    'bank',
    'cabar'
]
    # 'bat',
    # 'dog',
    # 'light',
    # 'plant',
    # 'book',
    # 'teacher',
    # 'student',
    # 'computer'
# ]

print("Selected words:")
print(words)
print("Total words:", len(words))
```

```python
# Display synsets for each word

for word in words:
    print("\nWord:", word)
    
    synsets = wn.synsets(word)
    
    for syn in synsets:
        print("  ", syn.name())
```

```python
# Display synset name, definition and example sentence

for word in words:
    print("\n" + "=" * 60)
    print("WORD:", word.upper())
    
    synsets = wn.synsets(word)
    
    # Display first 3 synsets for each word
    for syn in synsets[:3]:
        print("\nSynset:", syn.name())
        print("Definition:", syn.definition())
        
        if syn.examples():
            print("Example:", syn.examples()[0])
        else:
            print("Example: No example available")
```

```python
# Find words having multiple meanings

print("Words with multiple meanings:\n")

for word in words:
    synsets = wn.synsets(word)
    
    if len(synsets) > 1:
        print(word, "has", len(synsets), "synsets")
        
        # Display first 3 meanings
        for syn in synsets[:3]:
            print("  -", syn.name(), ":", syn.definition())
        
        print()
```

```python
import pandas as pd

data = []

for word in words:
    for syn in wn.synsets(word)[:3]:
        
        example = syn.examples()[0] if syn.examples() else "No example available"
        
        data.append([
            word,
            syn.name(),
            syn.definition(),
            example
        ])

q1_table = pd.DataFrame(
    data,
    columns=["Word", "Synset", "Definition", "Example"]
)

display(q1_table)
```

```python
# Q2(a): Select at least 10 nouns

nouns = [
    'dog',
    'cat',
    'car',
    'rose',
    'apple',
    'computer',
    'teacher',
    'student',
    'bird',
    'fish'
]

print("Selected nouns:")
print(nouns)
print("Total nouns:", len(nouns))
```

```python
# Q2(b) and Q2(c)
# Find a suitable noun synset and its direct hypernyms

data = []

for word in nouns:
    
    # Get noun synsets
    synsets = wn.synsets(word, pos=wn.NOUN)
    
    if synsets:
        syn = synsets[0]
        
        # Get direct hypernyms
        hypernyms = syn.hypernyms()
        
        hypernym_names = [h.name() for h in hypernyms]
        
        data.append([
            word,
            syn.name(),
            ", ".join(hypernym_names)
        ])

q2_direct = pd.DataFrame(
    data,
    columns=["Word", "Synset", "Direct Hypernyms"]
)

display(q2_direct)
```

```python
# Q2(d): Function to trace multiple levels of hypernyms

def hypernym_chain(synset, levels=3):
    
    chain = [synset.name()]
    current = synset
    
    for i in range(levels):
        
        hypernyms = current.hypernyms()
        
        if not hypernyms:
            break
        
        current = hypernyms[0]
        chain.append(current.name())
    
    return " -> ".join(chain)
```

```python
# Trace hypernym hierarchy for at least 5 words

five_words = [
    'dog',
    'cat',
    'car',
    'rose',
    'apple'
]

for word in five_words:
    
    syn = wn.synsets(word, pos=wn.NOUN)[0]
    
    print("\n", word)
    print("Hierarchy:")
    print(hypernym_chain(syn, levels=3))
```

```python
# Q2(e): Create final DataFrame with hypernym chains

data = []

for word in five_words:
    
    syn = wn.synsets(word, pos=wn.NOUN)[0]
    
    chain = hypernym_chain(syn, levels=3)
    
    data.append([
        word,
        syn.name(),
        chain
    ])

q2_final = pd.DataFrame(
    data,
    columns=["Word", "Synset", "Hypernym Chain"]
)

display(q2_final)
```

```python
pairs = [
    ('car', 'automobile'),
    ('dog', 'cat'),
    ('dog', 'animal'),
    ('dog', 'computer'),
    ('teacher', 'student'),
    ('rose', 'flower'),
    ('car', 'bicycle'),
    ('computer', 'machine')
]

print("Selected word pairs:")

for pair in pairs:
    print(pair)
```

```python
for w1, w2 in pairs:
    
    syn1 = wn.synsets(w1, pos=wn.NOUN)[0]
    syn2 = wn.synsets(w2, pos=wn.NOUN)[0]
    
    print("\nPair:", w1, "-", w2)
    print("Synset 1:", syn1.name())
    print("Synset 2:", syn2.name())
```

```python
for w1, w2 in pairs:
    
    syn1 = wn.synsets(w1, pos=wn.NOUN)[0]
    syn2 = wn.synsets(w2, pos=wn.NOUN)[0]
    
    score = syn1.path_similarity(syn2)
    
    print(w1, "-", w2, "=", score)
```

```python
data = []

for w1, w2 in pairs:
    
    syn1 = wn.synsets(w1, pos=wn.NOUN)[0]
    syn2 = wn.synsets(w2, pos=wn.NOUN)[0]
    
    score = syn1.path_similarity(syn2)
    
    data.append([
        w1,
        w2,
        syn1.name(),
        syn2.name(),
        score
    ])

q3 = pd.DataFrame(
    data,
    columns=[
        "Word 1",
        "Word 2",
        "Synset 1",
        "Synset 2",
        "Similarity"
    ]
)

display(q3)
```

```python
q3_ranked = q3.sort_values(
    by="Similarity",
    ascending=False
).reset_index(drop=True)

q3_ranked.index = q3_ranked.index + 1

display(q3_ranked)
```

```python
required_pairs = [
    ('car', 'automobile'),
    ('dog', 'cat'),
    ('dog', 'animal'),
    ('dog', 'computer')
]

comparison = q3[
    q3.apply(
        lambda row: (row["Word 1"], row["Word 2"]) in required_pairs,
        axis=1
    )
]

display(comparison)
```
