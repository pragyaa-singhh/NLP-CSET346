```python
import re

def clean_text(text):

    # 1. Lowercase
    text = text.lower()

    # 2. Handle contractions
    contractions = {
        "can't": "cannot",
        "i'm": "i am",
        "it's": "it is",
        "i’ve": "i have",
        "i've": "i have",
        "can’t": "cannot",
        "what’s": "what is",
        "you're": "you are",
        "u're": "you are",
        "c'mon": "come on",
        "waz": "was",
        "wuz": "was",
        "cud": "could",
        "u ": "you ",
        "u're": "you are"
    }

    for contraction, replacement in contractions.items():
        text = text.replace(contraction, replacement)

    # I <3 NLP → I love NLP
    text = text.replace("<3", " love ")

    # 3. Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    # 4. Remove special characters
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)

    # 5. Reduce repeated letters
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)

    # 6. Fix extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()

    return text


cleaned_paragraphs = []

for paragraph in paragraphs:
    cleaned = clean_text(paragraph)
    cleaned_paragraphs.append(cleaned)

for i, text in enumerate(cleaned_paragraphs, 1):
    print(f"\nParagraph {i}:")
    print(text)
```

```python
import nltk

nltk.download("brown")
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("wordnet")
nltk.download("omw-1.4")

print("NLTK resources downloaded!")
```

```python
from nltk.corpus import brown
from nltk.tokenize import word_tokenize

# Select category
category = "news"

words = brown.words(categories=category)

text = " ".join(words)

tokens = word_tokenize(text)

print("Category:", category)
print("Number of words:", len(words))
print("Number of tokens:", len(tokens))

print("\nFirst 50 tokens:")
print(tokens[:50])
```

```python
from nltk.probability import FreqDist

# Keep alphabetic words
word_tokens = [
    word.lower()
    for word in tokens
    if word.isalpha()
]

fdist = FreqDist(word_tokens)

print("Top 20 words:")

for word, frequency in fdist.most_common(20):
    print(word, ":", frequency)
```

```python
top20 = fdist.most_common(20)

words_20 = [x[0] for x in top20]
freq_20 = [x[1] for x in top20]

plt.figure(figsize=(12,6))

plt.bar(words_20, freq_20)

plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("Top 20 Word Frequencies - Brown Corpus")

plt.xticks(rotation=45)

plt.show()
```

```python
from nltk.stem import PorterStemmer
from collections import Counter

stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word)
    for word in word_tokens
]

stem_frequency = Counter(stemmed_words)

print("Top 20 Stemmed Words:\n")

for word, frequency in stem_frequency.most_common(20):
    print(word, ":", frequency)
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in word_tokens
]

lemma_frequency = Counter(lemmatized_words)

print("Top 20 Lemmatized Words:\n")

for word, frequency in lemma_frequency.most_common(20):
    print(word, ":", frequency)
```

```python
comparison_words = word_tokens[:15]

comparison = []

for word in comparison_words:

    stem = stemmer.stem(word)
    lemma = lemmatizer.lemmatize(word)

    comparison.append([
        word,
        stem,
        lemma
    ])

comparison_df = pd.DataFrame(
    comparison,
    columns=["Original Word", "Stemmed Word", "Lemmatized Word"]
)

comparison_df
```
