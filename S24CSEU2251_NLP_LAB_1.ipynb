```python
# NLP LAB - TEXT PREPROCESSING AND N-GRAM ANALYSIS

import nltk
import string
import re
import pandas as pd

from nltk.corpus import brown, stopwords, words
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.util import ngrams
from nltk.probability import FreqDist
from difflib import get_close_matches

# Download required NLTK resources
resources = [
    'brown',
    'punkt',
    'punkt_tab',
    'stopwords',
    'wordnet',
    'omw-1.4',
    'words'
]

for resource in resources:
    nltk.download(resource, quiet=True)

print("All required libraries and NLTK resources are ready.")
```

```python
# Q1(a): Load Brown Corpus and select at least 20 sentences

brown_sentences = brown.sents()[:20]

# Convert word lists into sentences
sentences = [' '.join(sentence) for sentence in brown_sentences]

print("Total sentences selected:", len(sentences))

print("\nOriginal Brown Corpus Sentences:\n")

for i, sentence in enumerate(sentences[:10], 1):
    print(f"{i}. {sentence}")
```

```python
# Q1(b): Perform sentence-level and word-level tokenization

# Combine selected sentences into one text
text = ' '.join(sentences)

# Sentence tokenization
sentence_tokens = sent_tokenize(text)

print("Sentence Tokenization:\n")

for i, sentence in enumerate(sentence_tokens[:5], 1):
    print(f"{i}. {sentence}")

# Word tokenization
word_tokens = [word_tokenize(sentence) for sentence in sentence_tokens]

print("\nWord Tokenization:\n")

for i, tokens in enumerate(word_tokens[:5], 1):
    print(f"Sentence {i}: {tokens}")
```

```python
# Q1(c) and Q1(d):
# Convert tokens to lowercase
# Remove punctuation
# Remove English stop words

stop_words = set(stopwords.words('english'))

# Convert to lowercase
lower_tokens = [
    [word.lower() for word in sentence]
    for sentence in word_tokens
]

# Remove punctuation and stop words
cleaned_tokens = [
    [
        word for word in sentence
        if word.isalpha() and word not in stop_words
    ]
    for sentence in lower_tokens
]

print("Lowercase and cleaned tokens:\n")

for i in range(5):
    print(f"\nSentence {i+1}")
    print("Lowercase:", lower_tokens[i])
    print("Cleaned  :", cleaned_tokens[i])
```

```python
# Q1(e): Display original and cleaned tokens for at least five sentences

comparison_q1 = pd.DataFrame({
    'Sentence No.': range(1, 6),
    'Original Tokens': lower_tokens[:5],
    'Cleaned Tokens': cleaned_tokens[:5]
})

display(comparison_q1)

print("\nQ1 completed successfully.")
```

```python
# Q2(a): Extract at least 30 alphabetic words from Brown Corpus

brown_words = [
    word.lower()
    for word in brown.words()
    if word.isalpha()
]

# Required words from the question
required_words = [
    'playing',
    'studies',
    'running',
    'better',
    'cars'
]

# Take 30 Brown Corpus words
selected_words = brown_words[:30]

# Add required words
selected_words.extend(required_words)

# Remove duplicates while maintaining order
selected_words = list(dict.fromkeys(selected_words))

print("Total words selected:", len(selected_words))

print("\nSelected Words:")
print(selected_words)
```

```python
# Q2(b) and Q2(c):
# Apply Porter Stemmer and WordNet Lemmatizer

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Porter stemming
stems = [
    stemmer.stem(word)
    for word in selected_words
]

# POS information for required examples
pos_mapping = {
    'playing': 'v',
    'studies': 'v',
    'running': 'v',
    'better': 'a',
    'cars': 'n'
}

# Lemmatization function
def get_lemma(word):
    return lemmatizer.lemmatize(
        word,
        pos=pos_mapping.get(word, 'n')
    )

lemmas = [
    get_lemma(word)
    for word in selected_words
]

# Create comparison table
comparison_q2 = pd.DataFrame({
    'Original Word': selected_words,
    'Stem': stems,
    'Lemma': lemmas
})

display(comparison_q2)
```

```python
# Q2(d): Check required words
# playing, studies, running, better and cars

print("Required Example Words:\n")

required_comparison = comparison_q2[
    comparison_q2['Original Word'].isin(required_words)
]

display(required_comparison)

print("\nObservation:")
print("Stemming removes word endings mechanically,")
print("while lemmatization converts words to meaningful dictionary forms.")
```

```python
# Q3(a): Load NLTK Words Corpus

vocabulary = sorted({
    word.lower()
    for word in words.words()
    if word.isalpha()
})

print("Total vocabulary words:", len(vocabulary))

# Q3(b): Create at least 10 intentionally misspelled words

misspelled_words = [
    'recieve',
    'definately',
    'seperate',
    'occured',
    'enviroment',
    'becuase',
    'wierd',
    'adress',
    'tommorow',
    'goverment'
]

print("\nIntentionally Misspelled Words:")

for word in misspelled_words:
    print(word)
```

```python
# Q3(c): Use approximate string matching

def suggest_correction(word, cutoff=0.60):
    """
    Find the closest word from the NLTK Words Corpus.
    """
    matches = get_close_matches(
        word.lower(),
        vocabulary,
        n=1,
        cutoff=cutoff
    )
    
    return matches[0] if matches else "No suggestion"


# Generate corrections
suggestions = [
    suggest_correction(word)
    for word in misspelled_words
]

# Q3(d): Display misspelled words and suggestions

spelling_df = pd.DataFrame({
    'Misspelled Word': misspelled_words,
    'Suggested Correction': suggestions
})

display(spelling_df)
```

```python
# Q3(e): Correct at least five example sentences

example_sentences = [
    "I definately want to visit the museum tommorow.",
    "She recieved the adress yesterday.",
    "The enviroment is very important for everyone.",
    "He seperated the papers becuase they were mixed.",
    "The goverment made a wierd decision."
]


def correct_sentence(sentence):
    """
    Correct misspelled alphabetic words
    while keeping punctuation.
    """
    
    tokens = re.findall(
        r"[A-Za-z]+|[^A-Za-z\s]",
        sentence
    )
    
    corrected_tokens = []
    
    for token in tokens:
        
        if token.isalpha():
            lower_token = token.lower()
            
            # Check whether the word exists
            if lower_token not in vocabulary:
                
                suggestion = suggest_correction(
                    lower_token
                )
                
                if suggestion != "No suggestion":
                    
                    # Preserve first-letter capitalization
                    if token[0].isupper():
                        suggestion = suggestion.capitalize()
                    
                    token = suggestion
        
        corrected_tokens.append(token)
    
    # Reconstruct sentence
    result = ""
    
    for token in corrected_tokens:
        
        if token in string.punctuation:
            result += token
            
        else:
            if result:
                result += " "
            
            result += token
    
    return result


# Correct all sentences
corrected_sentences = [
    correct_sentence(sentence)
    for sentence in example_sentences
]


# Display results
print("SPELLING CORRECTION OF SENTENCES")
print("=" * 70)

for i, (original, corrected) in enumerate(
    zip(example_sentences, corrected_sentences),
    1
):
    
    print(f"\nSentence {i}")
    print("Original :", original)
    print("Corrected:", corrected)

print("\nQ3 completed successfully.")
```

```python
# ============================================================
# Q4: UNIGRAM, BIGRAM AND TRIGRAM ANALYSIS
# ============================================================

# Q4(a): Load a suitable subset of Brown Corpus
brown_subset = brown.words()[:5000]

# Q4(b): Retain lowercase alphabetic tokens
preprocessed_tokens = [
    word.lower()
    for word in brown_subset
    if word.isalpha()
]

print("Number of original words:", len(brown_subset))
print("Number of preprocessed tokens:", len(preprocessed_tokens))

print("\nFirst 50 Preprocessed Tokens:")
print(preprocessed_tokens[:50])


# Q4(c): Generate N-grams

unigrams = list(ngrams(preprocessed_tokens, 1))
bigrams = list(ngrams(preprocessed_tokens, 2))
trigrams = list(ngrams(preprocessed_tokens, 3))


print("\nFirst 10 Unigrams:")
print(unigrams[:10])

print("\nFirst 10 Bigrams:")
print(bigrams[:10])

print("\nFirst 10 Trigrams:")
print(trigrams[:10])


# Q4(d): Calculate N-gram frequencies

unigram_freq = FreqDist(unigrams)
bigram_freq = FreqDist(bigrams)
trigram_freq = FreqDist(trigrams)


print("\nUnique Unigrams:", len(unigram_freq))
print("Unique Bigrams :", len(bigram_freq))
print("Unique Trigrams:", len(trigram_freq))
```

```python
# Q4(e): Display ten most frequent unigrams, bigrams and trigrams

# Top 10 unigrams
top_unigrams = pd.DataFrame(
    [
        (gram[0], frequency)
        for gram, frequency in unigram_freq.most_common(10)
    ],
    columns=['Unigram', 'Frequency']
)


# Top 10 bigrams
top_bigrams = pd.DataFrame(
    [
        (' '.join(gram), frequency)
        for gram, frequency in bigram_freq.most_common(10)
    ],
    columns=['Bigram', 'Frequency']
)


# Top 10 trigrams
top_trigrams = pd.DataFrame(
    [
        (' '.join(gram), frequency)
        for gram, frequency in trigram_freq.most_common(10)
    ],
    columns=['Trigram', 'Frequency']
)


print("=" * 70)
print("TOP 10 MOST FREQUENT UNIGRAMS")
print("=" * 70)
display(top_unigrams)


print("=" * 70)
print("TOP 10 MOST FREQUENT BIGRAMS")
print("=" * 70)
display(top_bigrams)


print("=" * 70)
print("TOP 10 MOST FREQUENT TRIGRAMS")
print("=" * 70)
display(top_trigrams)
```

```python
# Q4(f) Effect of Increasing N in N-grams

An N-gram is a sequence of N consecutive words.

### Unigram

A unigram contains only one word.

Example:

`machine`

It provides very little contextual information.

### Bigram

A bigram contains two consecutive words.

Example:

`machine learning`

It provides more information about the relationship between words.

### Trigram

A trigram contains three consecutive words.

Example:

`machine learning model`

It captures more contextual and word-order information.

### Effect of Increasing N

As N increases, the N-gram captures more context.

Therefore:

**Unigram → Less context**

**Bigram → More context**

**Trigram → Even more context**

However, increasing N also increases the number of possible word combinations. Higher-order N-grams occur less frequently, which can cause **data sparsity**.

Therefore, there is a trade-off between:

- More contextual information
- More data requirements
- Higher computational complexity
- Data sparsity

For this reason, the appropriate value of N depends on the application and the amount of available training data.
```
