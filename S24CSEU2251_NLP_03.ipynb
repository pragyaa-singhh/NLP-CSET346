```python
import nltk
import re
from collections import Counter
from nltk.util import ngrams

nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.corpus import brown

# Select a category
category = 'news'

# Get words from Brown Corpus
words = brown.words(categories=category)

# Take at least 5000 tokens
tokens = [word.lower() for word in words[:5000]]

# Basic preprocessing: keep alphabetic words
tokens = [word for word in tokens if word.isalpha()]

print("Number of tokens:", len(tokens))
print("First 20 tokens:", tokens[:20])
```

```python
total_tokens = len(tokens)
vocabulary = set(tokens)
vocabulary_size = len(vocabulary)

print("Total Tokens:", total_tokens)
print("Vocabulary Size:", vocabulary_size)
```

```python
unigrams = list(ngrams(tokens, 1))
bigrams = list(ngrams(tokens, 2))
trigrams = list(ngrams(tokens, 3))

unigram_freq = Counter(unigrams)
bigram_freq = Counter(bigrams)
trigram_freq = Counter(trigrams)

print("Number of Unigrams:", len(unigram_freq))
print("Number of Bigrams:", len(bigram_freq))
print("Number of Trigrams:", len(trigram_freq))
```

```python
print("Top 10 Unigrams:")
for item, count in unigram_freq.most_common(10):
    print(item, ":", count)

print("\nTop 10 Bigrams:")
for item, count in bigram_freq.most_common(10):
    print(item, ":", count)

print("\nTop 10 Trigrams:")
for item, count in trigram_freq.most_common(10):
    print(item, ":", count)
```

```python
print("Interpretation:")
print("1. Unigrams show the frequency of individual words.")
print("2. Bigrams show frequently occurring pairs of consecutive words.")
print("3. Trigrams show frequently occurring sequences of three words.")
print("4. Frequent n-grams help statistical language models learn common language patterns.")
```

```python
import nltk
import math
from collections import Counter
from nltk.util import ngrams

nltk.download('gutenberg')
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.corpus import gutenberg

# Select English text
words = gutenberg.words('austen-emma.txt')

# Normalize
tokens = [word.lower() for word in words if word.isalpha()]

print("Total tokens:", len(tokens))
print("Vocabulary size:", len(set(tokens)))
```

```python
unigram_counts = Counter(tokens)
bigram_counts = Counter(zip(tokens[:-1], tokens[1:]))

print("Total unique unigrams:", len(unigram_counts))
print("Total unique bigrams:", len(bigram_counts))

print("\nTop 10 Bigrams:")
for bigram, count in bigram_counts.most_common(10):
    print(bigram, ":", count)
```

```python
vocab_size = len(unigram_counts)

def bigram_probability(w1, w2):
    """
    Add-one Laplace smoothing:
    P(w2|w1) = (Count(w1,w2) + 1) / (Count(w1) + V)
    """
    numerator = bigram_counts[(w1, w2)] + 1
    denominator = unigram_counts[w1] + vocab_size

    return numerator / denominator
```

```python
def sentence_score(sentence):
    words = [word.lower() for word in sentence.split() if word.isalpha()]

    score = 0

    for i in range(len(words) - 1):
        prob = bigram_probability(words[i], words[i + 1])
        score += math.log(prob)

    return score


sentences = [
    "she was very happy",
    "the man went home",
    "emma was a young woman"
]

for sentence in sentences:
    print(sentence)
    print("Score:", sentence_score(sentence))
    print()
```

```python
print("Observations:")
print("1. Bigram models estimate the probability of a word based on the previous word.")
print("2. Add-one smoothing prevents zero probability for unseen bigrams.")
print("3. A higher sentence score indicates that the sentence is more probable according to the model.")
print("4. Smoothing is important because unseen word pairs can otherwise receive probability zero.")
```

```python
import nltk
import random

nltk.download('brown')
nltk.download('universal_tagset')

from nltk.corpus import brown

tagged_sentences = brown.tagged_sents(
    categories='news',
    tagset='universal'
)

# Convert CorpusView to list
tagged_sentences = list(tagged_sentences)

print("Total tagged sentences:", len(tagged_sentences))
print("Example:")
print(tagged_sentences[0])
```

```python
random.seed(42)
random.shuffle(tagged_sentences)

split = int(0.8 * len(tagged_sentences))

train_data = tagged_sentences[:split]
test_data = tagged_sentences[split:]

print("Training sentences:", len(train_data))
print("Testing sentences:", len(test_data))
```

```python
from nltk.tag import hmm

trainer = hmm.HiddenMarkovModelTrainer()

hmm_tagger = trainer.train_supervised(train_data)

print("HMM POS Tagger trained successfully.")
```

```python
test_sentences = [
    ["The", "company", "reported", "strong", "profits"],
    ["The", "boy", "is", "playing", "football"],
    ["She", "is", "reading", "a", "book"],
    ["The", "government", "announced", "new", "policies"],
    ["Students", "are", "learning", "natural", "language", "processing"]
]

for sentence in test_sentences:
    result = hmm_tagger.tag(sentence)
    print("Sentence:", " ".join(sentence))
    print("POS Tags:", result)
    print()
```

```python
accuracy = hmm_tagger.accuracy(test_data)

print("HMM POS Tagging Accuracy:", round(accuracy * 100, 2), "%")
```

```python
print("HMM Components:")
print("1. Hidden States: POS tags such as NOUN, VERB, ADJ, etc.")
print("2. Observations: Actual words present in the sentence.")
print("3. Transition Probability: Probability of one POS tag following another.")
print("4. Emission Probability: Probability of a word being generated by a particular POS tag.")
print("5. HMM predicts the most probable sequence of POS tags for the observed words.")
```

```python
from sklearn.datasets import fetch_20newsgroups
from collections import Counter
import re

categories = ['sci.space', 'comp.graphics']

newsgroups = fetch_20newsgroups(
    subset='train',
    categories=categories,
    remove=('headers', 'footers', 'quotes')
)

print("Number of documents:", len(newsgroups.data))
print("Categories:", newsgroups.target_names)
```

```python
# Combine all documents
text = " ".join(newsgroups.data).lower()

# Extract words
words = re.findall(r'\b[a-zA-Z]{3,}\b', text)

word_frequency = Counter(words)

print("Top 20 domain-specific terms:\n")

for word, count in word_frequency.most_common(20):
    print(word, ":", count)
```

```python
space_data = fetch_20newsgroups(
    subset='train',
    categories=['sci.space'],
    remove=('headers', 'footers', 'quotes')
)

graphics_data = fetch_20newsgroups(
    subset='train',
    categories=['comp.graphics'],
    remove=('headers', 'footers', 'quotes')
)

space_words = re.findall(
    r'\b[a-zA-Z]{3,}\b',
    " ".join(space_data.data).lower()
)

graphics_words = re.findall(
    r'\b[a-zA-Z]{3,}\b',
    " ".join(graphics_data.data).lower()
)

space_freq = Counter(space_words)
graphics_freq = Counter(graphics_words)

print("Top Sci.Space Terms:")
for word, count in space_freq.most_common(20):
    print(word, ":", count)

print("\nTop Comp.Graphics Terms:")
for word, count in graphics_freq.most_common(20):
    print(word, ":", count)
```

```python
print("Observations:")
print("1. Ambiguous words can have different meanings depending on surrounding context.")
print("2. Contextual clues help NLP systems determine the intended meaning.")
print("3. Coreference resolution identifies which entity a pronoun or referring expression represents.")
print("4. Different domains contain different vocabularies and frequently occurring terms.")
print("5. Domain-specific vocabulary can affect NLP performance because general language models may not know specialized meanings.")
```
